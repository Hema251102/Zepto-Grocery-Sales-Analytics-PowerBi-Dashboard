```python
 #Step 1: Load Data in Python
import pandas as pd  

# Load Excel file  
df = pd.read_excel("ZePtO Grocery Data.xlsx", sheet_name="Zepto")  
df.head()
```

```python
 #Step 2: Data Cleaning in Python
# Fix category labels
df["Item Fat Content"] = df["Item Fat Content"].replace({
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "reg": "Regular"
})
df["Item Fat Content"]
```

```python
df.nunique()  # Check unique values in each column
```

```python
df.isnull().sum()
```

```python
# Fill missing values
df["Item Weight"] = df["Item Weight"].fillna(df.groupby("Item Type")["Item Weight"].transform("mean"))
```

```python
df.isnull().sum()
```

```python
# Convert data types
df["Sales"] = df["Sales"].astype(float)
df["Rating"] = df["Rating"].astype(float)
df["Outlet Establishment Year"] = df["Outlet Establishment Year"].astype(int)
```

```python
# Create Outlet Age column
df["Outlet Age"] = 2024 - df["Outlet Establishment Year"]
```

```python
# Replace zero visibility with mean
df["Item Visibility"] = df["Item Visibility"].replace(0, df["Item Visibility"].mean())
```
